```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load cleaned data
df = pd.read_csv('../database/processed/sleep_data_clean.csv')

# Summary stats
print(df.describe())

# Histogram of sleep quality
sns.histplot(df['sleep_quality'], bins=10)
plt.title("Sleep Quality Distribution")
plt.savefig('../result/plot/sleep_quality_hist.png')
plt.show()

# Scatter: time in bed vs sleep quality
sns.scatterplot(data=df, x='time_in_bed_min', y='sleep_quality')
plt.title("Time in Bed vs Sleep Quality")
plt.savefig('../result/plot/time_vs_quality.png')
plt.show()

# Scatter: heart rate vs sleep quality
sns.scatterplot(data=df, x='heart_rate', y='sleep_quality')
plt.title("Heart Rate vs Sleep Quality")
plt.savefig('../result/plot/hr_vs_quality.png')
plt.show()
```
