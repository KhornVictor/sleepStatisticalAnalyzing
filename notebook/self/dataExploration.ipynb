## Histogram

```python
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
sns.histplot(time, kde=True, color='blue')
plt.title('Distribution of Time in Bed (minutes)')
plt.xlabel('Time in Bed (minutes)')
plt.ylabel('Frequency')
plt.subplot(1, 2, 2)
sns.histplot(heartRate, kde=True, color='green')
plt.title('Distribution of Heart Rate (bpm)')
plt.xlabel('Heart Rate (bpm)')
plt.ylabel('Frequency')
plt.tight_layout()
plt.savefig('../../result/plot/histogram.png')
plt.show()
```

## Box Plot

```python
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
sns.boxplot(y=time, color='blue')
plt.title('Boxplot of Time in Bed (minutes)')
plt.ylabel('Time in Bed (minutes)')
plt.subplot(1, 2, 2)
sns.boxplot(y=heartRate, color='green')
plt.title('Boxplot of Heart Rate (bpm)')
plt.ylabel('Heart Rate (bpm)')
plt.tight_layout()
plt.savefig('../../result/plot/boxplot.png')
plt.show()
```

## Kernel Density Estimation

```python
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
sns.kdeplot(time, fill=True, color='blue')
plt.title('KDE of Time in Bed (minutes)')
plt.xlabel('Time in Bed (minutes)')
plt.ylabel('Density')

plt.subplot(1, 2, 2)
sns.kdeplot(heartRate, fill=True, color='green')
plt.title('KDE of Heart Rate (bpm)')
plt.xlabel('Heart Rate (bpm)')
plt.ylabel('Density')

plt.tight_layout()
plt.savefig('../../result/plot/kde.png')
plt.show()
```

## Scatter plot

```python
plt.figure(figsize=(12, 5))
sns.scatterplot(x=time, y=heartRate)
plt.title('Scatter Plot of Time in Bed vs Heart Rate')
plt.xlabel('Time in Bed (minutes)')
plt.ylabel('Heart Rate (bpm)')
plt.savefig('../../result/plot/scatterplot.png')
plt.show()
```
