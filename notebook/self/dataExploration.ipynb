## Histogram

```python
plt.figure(figsize=(10, 5))
sns.histplot(time, kde=True, color='blue')
plt.title('Distribution of Time in Bed (minutes)')
plt.xlabel('Time in Bed (minutes)')
plt.ylabel('Frequency')
plt.tight_layout()
plt.savefig('../../result/plot/histogram.png')
plt.show()
```

## Box Plot

```python
plt.figure(figsize=(8, 10))
sns.boxplot(y=time, color='blue')
plt.title('Boxplot of Time in Bed (minutes)')
plt.ylabel('Time in Bed (minutes)')
plt.tight_layout()
plt.savefig('../../result/plot/boxplot.png')
plt.show()
```

## Kernel Density Estimation

```python
plt.figure(figsize=(10, 5))

sns.kdeplot(time, fill=True, color='blue')
plt.title('KDE of Time in Bed (minutes)')
plt.xlabel('Time in Bed (minutes)')
plt.ylabel('Density')

plt.tight_layout()
plt.savefig('../../result/plot/kde.png')
plt.show()
```
