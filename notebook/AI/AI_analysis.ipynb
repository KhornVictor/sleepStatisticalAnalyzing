```python
import pandas as pd
import statsmodels.api as sm

# Load data
df = pd.read_csv('../database/processed/sleep_data_clean.csv')

# Regression: sleep_quality ~ time_in_bed_min + heart_rate
X = df[['time_in_bed_min', 'heart_rate']]
y = df['sleep_quality']
X = sm.add_constant(X)
model = sm.OLS(y, X).fit()
print(model.summary())
```
